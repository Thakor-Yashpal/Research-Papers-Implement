```python
# Problem 1 — Remove duplicates while preserving order
nums = [1, 2, 3, 4, 5, 6, 99.1, 99.2, 6, 5, 4, 12, 3, 2, 1, 19, 233]

def Remove_duplicates(nums):
    seen = set()
    result = []
    for num in nums:
        if num not in seen:
            seen.add(num)
            result.append(num)
    return result

# Call the function and store/print the result correctly
unique_nums = Remove_duplicates(nums)

print(f"These are the unique elements in this list: {unique_nums}")
```

```python
# Problem 2 — First non-repeating character

# Implement first_unique_char(s) to return the first character that occurs exactly once, or None if there isn't one.
# Requirements: preserve character order, don't use Counter, aim for O(n) time.
# Concepts: Dictionaries, frequency counting, two-pass algorithms.

char = "kay karu hai kusch kusch hota hia ab to maera dill jage na so tah hai kay karu hia kuch kuch hota hia "


def first_unique_char(s):
    counts = {}

    for c in s:
        counts[c] = counts.get(c,0) + 1

    for c in s:
        if counts[c] == 1:
            return c

    return None

# Test the function
result = first_unique_char(char)
print(f"The first non-repeating character is: {repr(result)}")
```

```python
# Problem 3 — Two Sum
# (Brute Force - $O(n^2)$)
# Implement two_sum(nums, target) that returns the indices of two distinct elements whose sum equals target. Assume exactly one valid answer exists.

nums = (2, 7, 11, 1)
target = 12

def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return i, j  # Return indices, not values!

result = two_sum(nums, target)
print(f"The indices are: {result}") 


# The Optimal Solution ($O(n)$ Time Complexity)


nums = (2, 7, 11, 1)
target = 18

def twosum(nums, target):
    seen = {}  # Stores {number: index}
    
    for i, num in enumerate(nums):
        complement = target - num
        
        # If the complement is already in our dictionary, we found our pair!
        if complement in seen:
            return [seen[complement], i]
            
        # Otherwise, add the current number and its index to the dictionary
        seen[num] = i

result = twosum(nums, target)
print(f"The indices are: {result}")  # Output: [0, 3]
```

```python
# Problem 4 — Merge two sorted lists

# Implement merge_sorted(a, b) that returns one sorted list without calling sorted().

def merge_sorted(a, b):
    result = []
    i = 0  # Pointer for list a
    j = 0  # Pointer for list b
    
    # Compare elements from both lists and append the smaller one
    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1
            
    # If any elements remain in list a, append them
    while i < len(a):
        result.append(a[i])
        i += 1
        
    # If any elements remain in list b, append them
    while j < len(b):
        result.append(b[j])
        j += 1
        
    return result

# Test cases
print(merge_sorted([1, 3, 5], [2, 4, 6]))  # Output: [1, 2, 3, 4, 5, 6]
print(merge_sorted([], [1, 2]))           # Output: [1, 2]
```
