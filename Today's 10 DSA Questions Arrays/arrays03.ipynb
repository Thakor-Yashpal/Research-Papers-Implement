```python
# 1. Find the largest element

arr = [10, 25, 7, 45, 18]

largest = arr[0]

for num in arr:
    if num > largest:
        largest = num

print(largest)
```

```python
# 2. Find the smallest element

arr = [12, 5, 8, 20, 3, 15]

smallest = arr[0]

for num in arr:
    if num < smallest:
        smallest = num

print(smallest)
```

```python
# 3. Calculate the sum

arr = [10, 20, 30, 40, 50]

total = 0

for num in arr:
    total += num

print(total)
```

```python
# 4. Count even numbers
arr = [1, 4, 7, 10, 13, 16, 20]

count = 0

for num in arr:
    if num % 2 ==0:
        count += 1

print(count)
```

```python
# 5. Count odd numbers

arr = [3, 8, 11, 14, 17, 22]

count = 0

for num in arr:
    if num % 2 != 0:
        count += 1

print(count)
```

```python
# 6. Reverse an array'ArithmeticError

arr = [1, 2, 3, 4, 5]

reversed_arr = []

for i in range(len(arr) - 1, -1, -1):
    reversed_arr.append(arr[i])

print(reversed_arr)
```
