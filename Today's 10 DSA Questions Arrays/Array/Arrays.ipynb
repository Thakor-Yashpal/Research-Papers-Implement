```python
# Q1. Concatenation of Array
# Problem

# Given an integer array nums of length n, create an array ans of length 2n such that:
```

```python
class Solution:
    def getConcatenation(self, nums):
        # Write your code here
        n = len(nums)
        ans = [0] * (2 * n)

        for i in range(n):
            ans[i] = nums[i]
            ans[i + n] = nums[i]
        return ans

solution = Solution()
result = solution.getConcatenation([1, 2, 3])
print(result)  # Output: [1, 2, 3, 1, 2, 3]
```

```python
# Q2. Shuffle the Array

class Solution:
    def shuffle(self, nums, n):
        # Write your code here
        suffled = []
        for i in range(n):
            suffled.append(nums[i])
            suffled.append(nums[i + n])
        return suffled

solution = Solution()
result = solution.shuffle([2, 5, 1, 3, 4, 7], 3)
print(result)  # Output: [2, 3, 5, 4, 1, 7]
```

```python
# Q3. Max Consecutive Ones

# Given a binary array nums, return the maximum number of consecutive 1s.

class Solution:
    def findMaxConsecutiveOnes(self, nums):
        # Write your code here
        max_count = 0
        current_count = 0
        for num in nums:
            if num == 1:
                current_count += 1
                max_count = max(max_count, current_count)
            else:
                current_count = 0
        return max_count

solution = Solution()
result = solution.findMaxConsecutiveOnes([1, 1, 0, 1, 1, 1])
print(result)  # Output: 3
```

```python
# Q4. Contains Duplicate

# Given an integer array nums, return True if any value appears at least twice.

class Solution:
    def containsDuplicate(self, nums):
        # Write your code here
        seen = set()

        for num in nums:
            if num in seen:
                return True
            seen.add(num)
        return False

solution = Solution()
result = solution.containsDuplicate([1, 2, 3, 1])
print(result)  # Output: True
```

```python
# 🟢 Q5. Two Sum ⭐

# This is a very important interview question.

# Problem

# Given an array of integers nums and an integer target, return the indices of two numbers such that they add up to target.


nums = [2, 7, 11, 15]
target = 9

class Solution:
    def two_sum(self, nums, target):
        num_to_index = {}

        for index, num in enumerate(nums):
            complement = target - num
            if complement in num_to_index:
                return [num_to_index[complement], index]
            num_to_index[num] = index

solution = Solution()
result = solution.two_sum(nums, target)
print(result)  # Output: [0, 1]


# secod opction

class Solution:
    def two_sum(self, nums, target):
        for i in range(len(nums)):
            for j in range(i + 1, len(nums)):
                if nums[i] + nums[j] == target:
                    return [i, j]

solution = Solution()
result = solution.two_sum(nums, target) 
print(result)  # Output: [0, 1]
```

```python
# 🟢 Q6. Move Zeroes

# Given an integer array nums, move all 0s to the end while maintaining the relative order of non-zero elements.

# Example
# Input:  [0, 1, 0, 3, 12]

# Output: [1, 3, 12, 0, 0]

class Solution:
    def moveZeroes(self, nums):
        # Your code here
        last_non_zero_found_at = 0
        for i in range(len(nums)):
            if nums[i] != 0:
                nums[last_non_zero_found_at] = nums[i]
                last_non_zero_found_at += 1
        while last_non_zero_found_at < len(nums):
            nums[last_non_zero_found_at] = 0
            last_non_zero_found_at += 1
        return nums

solution = Solution()
result = solution.moveZeroes([0, 1, 0, 3, 12])
print(result)  # Output: [1, 3, 12, 0,
```

```python
# 🟢 Q7. Remove Duplicates from Sorted Array

# Given a sorted array nums, remove duplicates in-place such that each unique element appears only once.

# Example
# Input:  [1, 1, 2]

# Output: [1, 2]

# Another example:

# Input:  [0,0,1,1,1,2,2,3,3,4]

# Output: [0,1,2,3,4]

class Solution:
    def removeDuplicates(self, nums):

        if not nums:
            return 0

        write_index = 1
        for read_index in range(1,len(nums)):
            if nums[read_index] != nums[read_index - 1]:
                 nums[write_index] = nums[read_index]
                 write_index += 1
        return write_index

solution = Solution()
result = solution.removeDuplicates([0,0,1,1,1,2,2,3,3,4])
print(result)  # Output: 5 (The first five elements of nums are now [0, 1, 2, 3, 4])      
```

```python
# 🟢 Q8. Best Time to Buy and Sell Stock ⭐

# This is a very common interview question.

# Problem

# You are given:

# prices = [7,1,5,3,6,4]

# prices[i] represents the price of a stock on day i.

# You want to:

# Buy on one day.
# Sell on a future day.

# Return the maximum profit.

class Solution:
    def maxProfit(self, prices):

        min_price = float('inf')
        max_profit = 0

        for price in prices:
            if price < min_price:
                min_price = price

            elif price - min_price > max_profit:
                max_profit = price - min_price

        return max_profit


solution = Solution()
result = solution.maxProfit([7, 1, 5, 3, 6, 4])

print(result)  # Output: 5
```

```python
# Q9. Majority Element

# Given an array nums of size n, return the majority element.

# The majority element appears more than n / 2 times.

# Example 1
# nums = [3, 2, 3]

class Solution:
    def majorityElement(self, nums):
        count = {}

        for num in nums:
            count[num] = count.get(num, 0) + 1
            if count[num] > len(nums) // 2:
                return num

solution = Solution()
result = solution.majorityElement([3, 2, 3])
print(result)  # Output: 3
```

```python
# 🟢 Q10. Single Number

# Given a non-empty integer array nums, every element appears twice except for one element.

# Find that single element.

# Example 1
# nums = [2, 2, 1]

class Solution:
    def singleNumber(self, nums):
        result = 0
        for num in nums:
            result ^= num
        return result

solution = Solution()
result = solution.singleNumber([2, 2, 1])
print(result)  # Output: 1
```

```python
stringe = 'Hello, World! , Hello, World! , Hello, World! '

class Solution:
    def count_occurrences(self, stringe):

        words = stringe.split(', ')
        word_count = {}

        for word in words:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1
        return word_count

solution = Solution()
result = solution.count_occurrences(stringe)
print(result)  # Output: {'Hello, World!': 3}
```
