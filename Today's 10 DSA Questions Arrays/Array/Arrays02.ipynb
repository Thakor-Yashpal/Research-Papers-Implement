```python
# Q1. Concatenation of Array
# Problem

# Given an integer array nums of length n, create an array ans of length 2n such that:


class Solution:
    def getConcatenation(self, nums):

        n = len(nums)

        ans = [0] * (2 * n)

        for i in range(n):
            ans[i] = nums[i]
            ans[i + n] = nums[i]
        return ans


def main():
    nums = [1, 2, 3]
    solution = Solution()
    result = solution.getConcatenation(nums)
    print(result)

if __name__ == "__main__":
    main()
    
```

```python
# Q2. Shuffle the Array

class Solution:
    def shuffle(self, nums, n):
        suffled = []

        for i in range(n):
            suffled.append(nums[i])
            suffled.append(nums[i + n])
        return suffled

def main():
    nums = [2, 5, 1, 3, 4, 7]
    n = 3
    solution = Solution()
    result = solution.shuffle(nums, n)
    print(result)

if __name__ == "__main__":
    main()
```

```python
# Q3. Max Consecutive Ones

# Given a binary array nums, return the maximum number of consecutive 1s.


class Solution:
    def findMaxConsecutiveOnes(Self,nums):

        max_count = 0
        count = 0

        for num in nums:
            if num == 1:
                count += 1
                max_count = max(max_count, count)
            else:
                count = 0
        return max_count

def main():
    nums = [1, 1, 0, 1, 1, 1]
    solution = Solution()
    result = solution.findMaxConsecutiveOnes(nums)
    print(result)

if __name__ == "__main__":
    main()
```
