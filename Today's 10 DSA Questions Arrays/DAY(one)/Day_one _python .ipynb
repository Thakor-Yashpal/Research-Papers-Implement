````python
# Question 3 — Two Sum

# Given nums = [2, 7, 11, 15] and target = 9, return the indices of two numbers whose sum equals the target.

# Challenge: solve it in O(n) expected time using a dictionary.

nums = [2, 7, 11, 15]

target = 26

def two_sum1(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]

    raise ValueError("No two numbers sum to the target")


def two_sum2(nums, target):
    seen = {}
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
    return None

two_sum2(nums,target)

two_sum1(nums, target)

# Explanation: Store previously seen values and their indices. Look for the complement before storing the current number.
````

````python
# Question 5 — Binary search

# Implement binary search for a sorted list. Return the index of the target, or -1 if it is absent.

def binary_search(nums, target):
    lo, hi = 0, len(nums) - 1

    while lo <= hi:
        mid = (lo + hi) // 2
        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1

    return -1

binary_search(nums,target)
````

````python
# Q2. Remove duplicates from a list while preserving order.

def unique_items(nums):
    seen = set()
    result = []

    for num in nums:
        if num not in seen:
            seen.add(num)
            result.append(num)

    return result


print(unique_items([4, 2, 4, 1, 2, 5, 1]))

# Explanation:
# - A set tracks previously encountered elements.
# - A list preserves their original order.
# - Each element is processed once.
# # Complexity: O(n) expected time and O(n) extra space.
````

````python
# Q3. Explain the difference between is and ==.
# Answer:
# - == compares values for equality.
# - is checks whether two references point to the same object.
# Example:


a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)  # True
print(a is b)  # False
print(a is c)  # True
````

````python
# Q4. Write a function that returns the frequency of each word.
# Input:

text = "python torch python model torch python"


from collections import Counter

def word_frequency(text):
    return Counter(text.split())

print(word_frequency(
    "python torch python model torch python"
))


# we can also implement it manually:

def word_frequency2(text):
    counts = {}

    for word in text.split():
        counts[word] = counts.get(word, 0) + 1

    return counts

print(word_frequency2(
    "python torch python model torch python"
))


# Explanation: dict.get(word, 0) returns the existing count or zero if the word is new.

# Complexity: O(n) expected time, where n is the total number of characters processed.
````

````python
# Q5. What is the difference between a shallow copy and a deep copy?
# Answer:

# A shallow copy creates a new outer object but shares nested objects. A deep copy recursively copies nested objects.

import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 99

print(shallow)  # [[99, 2], [3, 4]]
print(deep)     # [[1, 2], [3, 4]]


# Explanation: The shallow copy shares the inner lists with original. The deep copy has independent nested lists in this example.

# ML engineering connection: Unexpected shared references can cause bugs when modifying nested configurations or data structures.
````
