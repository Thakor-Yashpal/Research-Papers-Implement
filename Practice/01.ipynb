```python
import torch 
import torch.nn as nn
```

```python
X = torch.randn(20,1)

y = torch.randn(20,1)
```

```python
class simple_model(nn.Module):
    def __init__(self,input_dim,hidden_01,hidden_02,hidden_03,hidden_04,output_dim):
        super().__init__()
        self.layer1 = nn.Linear(input_dim,hidden_01)
        self.layer2 = nn.Linear(hidden_01,hidden_02)
        self.layer3 = nn.Linear(hidden_02,hidden_03)
        self.layer4 = nn.Linear(hidden_03,hidden_04)
        self.layer5 = nn.Linear(hidden_04,output_dim)

    def forward(self,X):
        X = self.layer1(X)
        X = self.layer2(X)
        X = self.layer3(X)
        X = self.layer4(X)
        X = self.layer5(X)
        return X
```

```python
model = simple_model(
    input_dim=1,
    hidden_01=20,
    hidden_02=15,
    hidden_03=10,
    hidden_04=5,
    output_dim=1
)

prediction = model(X)

print("X shape:", X.shape)
print("Prediction shape:", prediction.shape)
```

```python
X = torch.randn(20, 1)


class Simple_model(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer01 = nn.Linear(1,10)
        self.layer02 = nn.Linear(10,5)
        self.layer03 = nn.Linear(5,1)

    def forward(self,X):
        X = torch.relu(self.layer01(X))
        X = torch.relu(self.layer02(X))
        X = self.layer03(X)

        return X

X = torch.randn(20,1)

model = Simple_model()

prediction = model(X)

print(prediction.shape)
```

```python
import torch
import torch.nn as nn

X = torch.randn(20, 1)
y = torch.randn(20, 1)

model = Simple_model()

loss_function = nn.MSELoss()

optimizer = torch.optim.SGD(model.parameters(),lr=0.01)

epochs = 1000

for epoch in range(epochs):

    prediction = model(X)

    loss = loss_function(prediction,y)

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch: {epoch + 1} | Loss: {loss.item():.4f}"
        )
```
