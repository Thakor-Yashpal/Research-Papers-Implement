```python
import torch 
import torch.nn as nn
```

```python
X = torch.randn(20,1)

y = torch.randn(20,1)
```

```python
class simple_model(nn.Module):
    def __init__(self,input_dim,hidden_01,hidden_02,hidden_03,hidden_04,output_dim):
        super().__init__()
        self.layer1 = nn.Linear(input_dim,hidden_01)
        self.layer2 = nn.Linear(hidden_01,hidden_02)
        self.layer3 = nn.Linear(hidden_02,hidden_03)
        self.layer4 = nn.Linear(hidden_03,hidden_04)
        self.layer5 = nn.Linear(hidden_04,output_dim)

    def forward(self,X):
        X = self.layer1(X)
        X = self.layer2(X)
        X = self.layer3(X)
        X = self.layer4(X)
        X = self.layer5(X)
```

```python
model = simple_model(
    input_dim=1,
    hidden_01=20,
    hidden_02=15,
    hidden_03=10,
    hidden_04=5,
    output_dim=1
)

prediction = model(X)
print(prediction.shape)
```
