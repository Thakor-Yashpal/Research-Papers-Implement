```python
import torch
import torch.nn as nn
```

```python
x = torch.rand(10,10)
y = torch.rand(20,20)
```

```python
class SimpleModel(nn.Module):
    def __init__(self, input_features, hidden_state01, hidden_state02,
                 hidden_state03, out_features):
        super().__init__()

        self.layer01 = nn.Linear(input_features, hidden_state01)
        self.layer02 = nn.Linear(hidden_state01, hidden_state02)
        self.layer03 = nn.Linear(hidden_state02, hidden_state03)
        self.layer04 = nn.Linear(hidden_state03, out_features)
        self.activation = nn.ReLU()

    def forward(self, x):
        x = self.activation(self.layer01(x))
        x = self.activation(self.layer02(x))
        x = self.activation(self.layer03(x))
        return self.layer04(x)


Model = SimpleModel(
    input_features=10,
    hidden_state01=20,
    hidden_state02=15,
    hidden_state03=10,
    out_features=1
)

output = Model(x)
print(output.shape)  # torch.Size([10, 1])
```

```python
s = "leagnth of the string sffsdffsdfsdfsdfsdfsdfsdfsd"

# Get the length of the string
length = len(s)
print(f"Length of the string: {length}")
```

```python

# Find duplicate characters and their frequencies
counts = {}

for char in s:
    counts[char] = counts.get(char, 0) + 1

duplicates = {
    char: count for char, count in counts.items() if count > 1
    }

print("Duplicate characters:", duplicates)
```

```python
text = "machine learning"

vowels = "aeiou"
count = 0

for char in text.lower():
    if char in vowels:
        count += 1

print(count)
```
